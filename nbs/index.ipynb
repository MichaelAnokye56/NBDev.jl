# NBJulia

> Notebook-driven development for Julia: write code, tests, and docs
> together in a notebook, export a real package, generate docs, run tests.
> Modeled on Jeremy Howard's nbdev.

````julia
#| hide
using NBJulia
````

## Install

```julia
using Pkg
Pkg.add(url="https://github.com/MichaelAnokye56/NBJulia.jl")
```

## How to use

See the [full README](https://github.com/MichaelAnokye56/NBJulia.jl#readme) for
the quickstart, directive syntax, command reference, and troubleshooting.
In short:

```julia
NBJulia.scaffold("MyLib"; dir = pwd(), user = "you", author = "You", author_email = "you@example.com")
NBJulia.export_project()   # write tagged notebook cells into src/
NBJulia.test_project()     # run every notebook as a test
NBJulia.render_docs()      # build a Quarto docs site
```
